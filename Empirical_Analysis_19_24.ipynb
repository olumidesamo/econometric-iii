### **DML Causal Inference Application : Impact of Armed Conflict of Food Production in Nigeria**

In this study, we implemented the DML approach to estimate the impact of armed conflict on quantity of food harvested (Output) in Kilogram by estimating the effect of number of conflict events and military presence on food production across different local goverments of Nigeria.
This study applied a different methodology (DML) to relevant study previously conducted `Adelaja & Weatherspoon (2019)`, `George & Adelaja (2022)`, `Kafando & Sakurai (2024)`, `Kafando & Sakurai (2025)`. <br>

Quantity of output is affected by many factors including assess to inputs, household income, weather conditions, market access and other idiosyncratic factors. However in the areas where activities of armed conflicts are prevalent, this could be a significant factors that determines area of land cultivated which can ultimately determine the output.  
There have been several report by rural dwellers in some aggrarian communtity where they report been attacked on their ways to farm, sometimes they are attacked on their farms and in extreme cases kidnapped. <br>

Another significant consideration is that intensity of armed conflict can lead to household displacement from their communities which is a prevailing event in many conflict ridden communities. In 2019, of the 33.41 million new IDPs, 24.85 million (74%) were displaced due to natural disasters while 8.55 million (26%) were displaced due to armed conflicts `(IDMC, 2021)`. All these contributes to the amount of area cultivated which consiquently affect the quantity of food produced. This motivates the idea of aggregating food production at local goverment (LGA) level. There are also reports that some LGAs has been overtaken by non-state conflict actors which includes the Lakurawa, Boko Haram, Islamic State of West Africa Province (ISWAP) <br>

In this study, we argue that number of conflict events do not randomly affect food production since conflict events have more than doubled in areas of high food production following the proposed bill of National Grazing Reserves Bill in 2016 which was later not ascented to by the National assembly due to the confidential nature of the bill. As land belonging to rural farmers will be reallocated to nomadic cattle herders. This makes conflict events endogenous to food production. Since military infrastructures where already situated before the conflict events more prominent particularly in the recent times since we used data from 2019 and 2024. Except for Borno state where more military were deployed to combat due to higher event of terror attacks. Most of the major sorghum producting states did not have military deployment except for the previous military bases(infrastructures). Therefore military base can be a strong instrument for conflict attacks since it only affects quantity of sorghum produced through conflict events.

**In the study military presence does not mean military camps were setup due to event of conflict, just like we purported above, only one of the states had a military camps setup because of the intensity of terror attacks in the states by Boko-Haram and ISWAP which is Borno State.** However for major highways in the states the are check points only on the roads. This a binary variable of 0 and 1 which implies no military presnce (infrastructures) and military presnce (infrastructures) respectively.

quant_harv_kg denote Quantity of harvest for this particular crop which can be be approximated as quantity of production. This was measured as a continuous variable in Kilograms

conflict_event refers to areas that have experienced event of conflicts. A dummy variable. where 1 = has experienced conflict event and 0 when there is no conflict event.
<br>

Why DML is important for this study previous studies have imposed a linear functional form on the effect of conflicts or terror attacks on food production including control variables `George & Adelaja (2022)`, `Kafando & Sakurai (2025)`. However this might be to strong an assumption to make, hence this neccessitate the application of DML which does not impose strong function form as theirs did.

## Polynomial Regression

Approximating $$g(X) \approx \beta_0 + \sum_{j=1}^{p} \beta_j x_j + \sum_{j=1}^{p} \sum_{k=j}^{p} \beta_{jk} x_j x_k + \sum_{j=1}^{p} \sum_{k=j}^{p} \sum_{m=k}^{p} \beta_{jkm} x_j x_k x_m + \cdots$$

The corresponding regression model：$$Y_i=D_i \alpha + \beta_0 + \sum_{j=1}^{p} \beta_j x_{ij} + \sum_{j=1}^{p} \sum_{k=j}^{p} \beta_{jk} x_{ij} x_{ik} + \epsilon_i$$

```python
from sklearn.preprocessing import PolynomialFeatures
import pandas as pd
import statsmodels.api as sm

features_for_poly = ['t2m', 'total_labor_day',
                   'rh2m', 'prectotcorr', 'inorg_fert_kg', 'org_fert_kg',
                   'pest_kg', 'herb_kg']

# Select the actual data from the DataFrame
X_data_for_poly = sorghum_data[features_for_poly]

poly = PolynomialFeatures(degree=2, include_bias=True, interaction_only=True)
feature_poly_array = poly.fit_transform(X_data_for_poly) # Pass the DataFrame subset, not the list of names
feature_names = poly.get_feature_names_out(input_features=X_data_for_poly.columns) # Use .columns from the DataFrame subset
# Create DataFrame with the original index from sorghum_data
feature_poly = pd.DataFrame(feature_poly_array, columns=feature_names, index=sorghum_data.index)
X_poly = pd.concat([feature_poly, sorghum_data[['military_presence']]], axis=1)

Y = sorghum_data['quant_harv_kg']

model = sm.OLS(Y,X_poly)
results_ols = model.fit()
print(results_ols.summary())
print('ATE：',results_ols.params.iloc[[-1]])
```

## Naive Machine Learning Implementation


# The Estimand

The estimand is:

$$
ATE = \mathbb{E}[Y(1) - Y(0)]
$$

where:

- $$Y(1) = \text{potential crop output if conflict occurs}$$

- $$Y(0) = \text{potential crop output if conflict does not occur}$$

In your application:

- **Outcome**

$$
Y = \texttt{quant_harv_kg}
$$

- **Treatment**

$$
D = \texttt{conflict_event}
$$

So the estimand is:

> the average causal effect of experiencing a conflict event on agricultural production.



# What Model is Being Estimated?

A fully nonparametric causal model estimated through DML-AIPW. Which was implemented via three nuisance functions.

---

## Outcome Regression Functions

$$
\mu_1(X) = \mathbb{E}[Y \mid D = 1, X]
$$

$$
\mu_0(X) = \mathbb{E}[Y \mid D = 0, X]
$$

These are estimated separately for treated and untreated groups.

---

## Propensity Score Model

$$
e(X) = P(D = 1 \mid X)
$$

estimated using:

- Logistic Lasso
- Random Forest Classifier
- Neural Network Classifier


The estimator is the AIPW score:

$$
\phi_i
=
\mu_1(X_i)
-
\mu_0(X_i)
+
\frac{
D_i \left(Y_i - \mu_1(X_i)\right)
}{
e(X_i)
}
-
\frac{
(1 - D_i)\left(Y_i - \mu_0(X_i)\right)
}{
1 - e(X_i)
}
$$

Then:

$$
\widehat{ATE}
=
\frac{1}{n}
\sum_{i=1}^{n}
\phi_i
$$

This is the classic:

- doubly robust,
- orthogonalized,
- cross-fitted DML estimator.


The essence of the "Double Machine Learning" was

Because the nuisance functions:

- $$\mu_1(X)$$
- $$\mu_0(X)$$
- $$e(X)$$

are estimated using machine learning methods such as:

- Lasso
- Random Forest
- Neural Networks

combined with:

- sample splitting,
- cross-fitting,
- orthogonal scores.

### **NOTE: This was not demeaned by taking into account the two-fixed effects from LGA and Year**

```python
import warnings
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import KFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GroupKFold


from sklearn.linear_model import LassoCV, LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.neural_network import MLPRegressor, MLPClassifier
from sklearn.exceptions import ConvergenceWarning


warnings.filterwarnings("ignore", category=ConvergenceWarning)
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)


# ------------------------------------------------------------
# 1. Load data
# ------------------------------------------------------------

# Outcome and treatment
if __name__ == "__main__":

    #
    features_base = ['field_size', 'total_labor_day', 'ha_harvest', 't2m',
                 'rh2m', 'prectotcorr', 'inorg_fert_kg', 'org_fert_kg',
                 'pest_kg', 'herb_kg']

    cols_needed = ["quant_harv_kg", "military_presence", 'conflict_event', 'lga', 'year'] + features_base

    # Keep only complete cases
    df = sorghum_data[cols_needed].dropna().copy()

    # Outcome
    Y = df["quant_harv_kg"].to_numpy(dtype=float)

    # Treatment
    Z = df["conflict_event"].to_numpy(dtype=int)

    # Covariates
    X = df[features_base].to_numpy(dtype=float)

    # Groups for GroupKFold
    groups = (df['lga'].astype(str) + '_' + df['year'].astype(str)).to_numpy()

# ------------------------------------------------------------
# 2. Learner factory
# ------------------------------------------------------------
def make_learners(model_type="lasso", random_state=123):
    """
    Returns:
        reg: regression learner for mu_0 and mu_1
        clf: classification learner for e(x)
    """

    if model_type == "lasso":
        reg = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LassoCV(cv=5, random_state=random_state, max_iter=100000))
        ])

        clf = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(
                penalty="l1",
                solver="liblinear",
                C=0.1,
                max_iter=100000,
                random_state=random_state
            ))
        ])

    elif model_type == "rf":
        reg = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", RandomForestRegressor(
                n_estimators=30000,
                min_samples_leaf=5,
                random_state=random_state,
                n_jobs=-1
            ))
        ])

        clf = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", RandomForestClassifier(
                n_estimators=30000,
                min_samples_leaf=5,
                random_state=random_state,
                n_jobs=-1
            ))
        ])

    elif model_type == "mlp":
        reg = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", MLPRegressor(
                hidden_layer_sizes=(64, 32),
                activation="relu",
                alpha=1e-3,
                max_iter=150000,
                random_state=random_state
            ))
        ])

        clf = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", MLPClassifier(
                hidden_layer_sizes=(64, 32),
                activation="relu",
                alpha=1e-3,
                max_iter=150000,
                random_state=random_state
            ))
        ])

    else:
        raise ValueError("model_type must be one of: 'lasso', 'rf', 'mlp'")

    return reg, clf


# ------------------------------------------------------------
# 3. Fully nonparametric DML-AIPW for ATE
# ------------------------------------------------------------
def dml_aipw_conf(Y, Z, X, groups, K=5, learner="lasso", epsilon_clip=0.01, random_state=123):
    """
    Cross-fitted DML-AIPW estimator for the ATE.

    Parameters
    ----------
    Y : array-like, shape (n,)
        Outcome.
    Z : array-like, shape (n,)
        Binary treatment.
    X : array-like, shape (n, p)
        Covariates.
    K : int
        Number of folds.
    learner : str
        'lasso', 'rf', or 'mlp'.
    epsilon_clip : float
        Propensity clipping threshold.
    random_state : int
        Random seed.

    Returns
    -------
    dict with:
        ate, se, ci_low, ci_high,
        mu0_hat, mu1_hat, e_hat, phi
    """

    Y = np.asarray(Y, dtype=float)
    Z = np.asarray(Z, dtype=int)
    X = np.asarray(X, dtype=float)

    n = len(Y)

    mu0_hat = np.zeros(n)
    mu1_hat = np.zeros(n)
    e_hat = np.zeros(n)

    gfk = GroupKFold(n_splits=K, shuffle=True, random_state=random_state)
    reg_base, clf_base = make_learners(learner, random_state=random_state)

    for train_idx, test_idx in gfk.split(X, groups=groups):
        X_train, X_test = X[train_idx], X[test_idx]
        Y_train, Z_train = Y[train_idx], Z[train_idx]

        # Outcome model for treated
        reg1 = clone(reg_base)
        reg1.fit(X_train[Z_train == 1], Y_train[Z_train == 1])

        # Outcome model for untreated
        reg0 = clone(reg_base)
        reg0.fit(X_train[Z_train == 0], Y_train[Z_train == 0])

        # Propensity model
        clf = clone(clf_base)
        clf.fit(X_train, Z_train)

        # Out-of-fold predictions
        mu1_hat[test_idx] = reg1.predict(X_test)
        mu0_hat[test_idx] = reg0.predict(X_test)
        e_hat[test_idx] = clf.predict_proba(X_test)[:, 1]

    # Clip propensity scores for numerical stability
    e_hat = np.clip(e_hat, epsilon_clip, 1 - epsilon_clip)

    # AIPW scores
    phi = (
        mu1_hat - mu0_hat
        + Z * (Y - mu1_hat) / e_hat
        - (1 - Z) * (Y - mu0_hat) / (1 - e_hat)
    )

    ate = phi.mean()
    psi = phi - ate
    se = np.sqrt(np.mean(psi ** 2) / n)

    ci_low = ate - 1.96 * se
    ci_high = ate + 1.96 * se

    return {
        "ate": ate,
        "se": se,
        "ci_low": ci_low,
        "ci_high": ci_high,
        "mu0_hat": mu0_hat,
        "mu1_hat": mu1_hat,
        "e_hat": e_hat,
        "phi": phi
    }


# ------------------------------------------------------------
# 4. Partially linear DML
# ------------------------------------------------------------
def dml_plr_lasso(Y, D, X, K=5, random_state=123):
    """
    Partially Linear Regression DML :
        Y = theta * D + g(X) + U
        D = m(X) + V
    estimated by cross-fitted orthogonal score with lasso/logit-lasso.
    """

    Y = np.asarray(Y, dtype=float)
    D = np.asarray(D, dtype=int)
    X = np.asarray(X, dtype=float)
    n = len(Y)

    ghat = np.zeros(n)
    mhat = np.zeros(n)

    reg_y = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", LassoCV(cv=5, random_state=random_state, max_iter=100000))
    ])

    reg_d = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            penalty="l1",
            solver="liblinear",
            C=0.1,
            max_iter=100000,
            random_state=random_state
        ))
    ])

    gkf = GroupKFold(n_splits=K, shuffle=True, random_state=random_state)

    for train_idx, test_idx in gkf.split(X, groups=groups): # Corrected line
        ry = clone(reg_y).fit(X[train_idx], Y[train_idx])
        rd = clone(reg_d).fit(X[train_idx], D[train_idx])

        ghat[test_idx] = ry.predict(X[test_idx])
        mhat[test_idx] = rd.predict_proba(X[test_idx])[:, 1]

    vhat = D - mhat
    theta = np.mean(vhat * (Y - ghat)) / np.mean(vhat * D)

    score = vhat * (Y - ghat - theta * (D - mhat))
    se = np.sqrt(np.mean(score ** 2) / (np.mean(vhat * D) ** 2 * n))

    return {
        "ate": theta,
        "se": se,
        "ci_low": theta - 1.96 * se,
        "ci_high": theta + 1.96 * se
    }


# ------------------------------------------------------------
# 5. Post-Double-Selection
# ------------------------------------------------------------

def post_double_selection_lasso(Y, D, X, random_state=123):
    """
    Simple PDS benchmark:
    1. Lasso of Y on X
    2. Logit-lasso of D on X
    3. Union of selected covariates
    4. OLS of Y on D and selected X
    """

    Y = np.asarray(Y, dtype=float)
    D = np.asarray(D, dtype=int)
    X = np.asarray(X, dtype=float)
    n = len(Y)

    scaler = StandardScaler()
    Xs = scaler.fit_transform(X)

    lasso_y = LassoCV(cv=5, random_state=random_state, max_iter=1000000).fit(Xs, Y)
    logit_d = LogisticRegression(
        penalty="l1",
        solver="liblinear",
        C=0.1,
        max_iter=1000000,
        random_state=random_state
    ).fit(Xs, D)

    sel_y = np.where(np.abs(lasso_y.coef_) > 1e-8)[0]
    sel_d = np.where(np.abs(logit_d.coef_.ravel()) > 1e-8)[0]
    sel = np.union1d(sel_y, sel_d)

    X_sel = X[:, sel]
    X_ols = np.column_stack([np.ones(n), D, X_sel])

    beta = np.linalg.lstsq(X_ols, Y, rcond=None)[0]
    resid = Y - X_ols @ beta

    # HC0 robust variance
    XtX_inv = np.linalg.inv(X_ols.T @ X_ols)
    meat = X_ols.T @ np.diag(resid ** 2) @ X_ols
    vcov = XtX_inv @ meat @ XtX_inv

    ate = beta[1]
    se = np.sqrt(vcov[1, 1])

    return {
        "ate": ate,
        "se": se,
        "ci_low": ate - 1.96 * se,
        "ci_high": ate + 1.96 * se,
        "n_selected_controls": len(sel)
    }


# ------------------------------------------------------------
# 6. Run all models
# ------------------------------------------------------------
res_dml_lasso = dml_aipw_conf(Y, Z, X, groups, K=5, learner="lasso", epsilon_clip=0.01, random_state=123)
res_dml_rf    = dml_aipw_conf(Y, Z, X, groups, K=5, learner="rf",    epsilon_clip=0.01, random_state=123)
res_dml_mlp   = dml_aipw_conf(Y, Z, X, groups, K=5, learner="mlp",   epsilon_clip=0.10, random_state=123)


results = pd.DataFrame([
    {
        "method": "DML-AIPW (Lasso)",
        "ATE": res_dml_lasso["ate"],
        "SE": res_dml_lasso["se"],
        "CI Low": res_dml_lasso["ci_low"],
        "CI High": res_dml_lasso["ci_high"],
    },
    {
        "method": "DML-AIPW (Random Forest)",
        "ATE": res_dml_rf["ate"],
        "SE": res_dml_rf["se"],
        "CI Low": res_dml_rf["ci_low"],
        "CI High": res_dml_rf["ci_high"],
    },
    {
        "method": "DML-AIPW (Neural Network)",
        "ATE": res_dml_mlp["ate"],
        "SE": res_dml_mlp["se"],
        "CI Low": res_dml_mlp["ci_low"],
        "CI High": res_dml_mlp["ci_high"],
    },

])

print(results.round(3))
```

# Partially Linear IV (PLIV) Model and DML Estimation of LATE



Here we estimate the Local Average Treatment Effect (LATE) using the Partially Linear Instrumental Variable Double Machine Learning (PLIV-DML) framework.<br>

### Parameter of Interest: ATE

The **Local Average Treatment Effect (LATE)** is

# Empirical Setup

- **Outcome Variable**

$$
Y = \texttt{quant_harv_kg}
$$

(crop production / harvested quantity)


- **Treatment Variable**

$$
D = \texttt{conflict_event}
$$

(binary indicator of conflict occurrence)


- **Instrumental Variable**

$$
Z = \texttt{military_presence}
$$

- **Control Variables**

$$
X = \texttt{weather, labor, fertilizer, field size, etc.}
$$

---

- **Fixed Effects**

Two-way fixed effects are removed using demeaning over:

- LGA fixed effects
- Year fixed effects

via:
$$
x_{it} - \bar{x}_{i} - \bar{x}_{t} + \bar{x}
$$

The **Partially Linear IV Model (PLIV)** was  estimated:

$$
Y = \theta D + g(X) + \varepsilon
$$

with endogenous treatment:

$$
D = r(X) + v
$$

and instrument equation:

$$
Z = m(X) + u
$$

where:

- $$g(X) = \mathbb{E}[Y \mid X]$$

- $$r(X) = \mathbb{E}[D \mid X]$$

- $$m(X) = \mathbb{E}[Z \mid X]$$

are estimated flexibly using machine learning methods such as:

- Lasso
- Random Forest
- Neural Networks


# The LATE


$$
\theta =
\frac{
\mathbb{E}\left[(Z - m(X))(Y - g(X))\right]
}{
\mathbb{E}\left[(Z - m(X))(D - r(X))\right]
}
$$

Was the orthogonalized IV estimand used in the DML-PLIV.


$$
\theta =
\frac{
\mathbb{E}[\tilde{Z}\tilde{Y}]
}{
\mathbb{E}[\tilde{Z}\tilde{D}]
}
$$

where:

$$
\tilde{Y} = Y - \hat{g}(X)
$$

$$
\tilde{D} = D - \hat{r}(X)
$$

$$
\tilde{Z} = Z - \hat{m}(X)
$$

These are residualized (orthogonalized) variables after partialling out \(X\).


# Finally

A partially linear IV model with two-way fixed effects was estimated:

$$
Y_{it} = \theta D_{it} + g(X_{it}) + \alpha_i + \lambda_t + \varepsilon_{it}
$$

where:

- $$\alpha_i = \text{LGA fixed effects}$$

- $$\lambda_t = \text{Year fixed effects}$$

implemented through demeaning.

```python
import warnings
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.model_selection import KFold
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Lasso, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.neural_network import MLPRegressor, MLPClassifier
from sklearn.exceptions import ConvergenceWarning
from sklearn.model_selection import GroupKFold
import pandas as pd


warnings.filterwarnings("ignore", category=ConvergenceWarning)
warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", category=UserWarning)



# ============================================================
# Demeaning
# ============================================================

def two_way_demean(df, cols, unit_col="lga", time_col="year"):
    """
    Removes unit fixed effects and time fixed effects:
    x_it - x_i_bar - x_t_bar + x_bar
    """
    out = df.copy()

    for col in cols:
        unit_mean = out.groupby(unit_col)[col].transform("mean")
        time_mean = out.groupby(time_col)[col].transform("mean")
        overall_mean = out[col].mean()

        out[f"{col}_twfe"] = out[col] - unit_mean - time_mean + overall_mean

    return out

# ---------------------------------------------------
# 1. Load data
# ---------------------------------------------------

if __name__ == "__main__":

    # Select variables needed for Part 1
    features_base = [
    'field_size', 'total_labor_day', 'ha_harvest', 't2m',
    'rh2m', 'prectotcorr', 'inorg_fert_kg', 'org_fert_kg',
    'pest_kg', 'herb_kg'
]

cols_needed = [
    "quant_harv_kg",
    "treatment_fatalities",
    'conflict_event',
    "military_presence",
    "lga",
    "year"
] + features_base


df = sorghum_data[cols_needed].dropna().copy()

cols_to_demean = [
    "quant_harv_kg",
    "treatment_fatalities",
    'conflict_event',
    "military_presence"
] + features_base

df = two_way_demean(
    df,
    cols=cols_to_demean,
    unit_col="lga",
    time_col="year"
)

Y = df["quant_harv_kg"].to_numpy(dtype=float)

D = df["conflict_event"].to_numpy(dtype=int)

Z = df["military_presence"].to_numpy(dtype=int)

X_cols_twfe = [f"{x}_twfe" for x in features_base]
X = df[X_cols_twfe].to_numpy(dtype=float)

groups = (df['lga'].astype(str) + '_' + df['year'].astype(str)).to_numpy() # Re-defining groups as 1D array of strings


# ---------------------------------------------------
# 2. Learner factory
# ---------------------------------------------------
def make_learners(model_type="lasso", random_state=123):
    if model_type == "lasso":
        # fixed-penalty lasso for speed/stability; replace by LassoCV if preferred
        reg = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", Lasso(alpha=0.001, max_iter=500000, random_state=random_state))
        ])

        clf = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", LogisticRegression(
                penalty="l1",
                solver="liblinear",
                C=1.0,
                max_iter=200000,
                random_state=random_state
            ))
        ])

    elif model_type == "rf":
        reg = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", RandomForestRegressor(
                n_estimators=2000,
                min_samples_leaf=10,
                random_state=random_state,
                n_jobs=-1
            ))
        ])

        clf = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("model", RandomForestClassifier(
                n_estimators=2000,
                min_samples_leaf=10,
                random_state=random_state,
                n_jobs=-1
            ))
        ])

    elif model_type == "mlp":
        reg = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", MLPRegressor(
                hidden_layer_sizes=(16, 8),
                activation="relu",
                alpha=1e-2,
                max_iter=1500,
                random_state=random_state
            ))
        ])

        clf = Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler()),
            ("model", MLPClassifier(
                hidden_layer_sizes=(64, 32),
                activation="relu",
                alpha=1e-3,
                max_iter=150000,
                random_state=random_state
            ))
        ])

    else:
        raise ValueError("model_type must be 'lasso', 'rf', or 'mlp'.")

    return reg, clf


# ---------------------------------------------------
# 3. DML-PLIV estimator
# ---------------------------------------------------
def pliv_dml_late_conf(Y, D, Z, X, groups, K=5,
                       learner_g="lasso",
                       learner_r="lasso",
                       learner_m="lasso",
                       random_state=123):
    """
    DML estimator for the PLIV/LATE parameter:
        theta = E[(Z-m(X))(Y-g(X))] / E[(Z-m(X))(D-r(X))]
    where
        g(X) = E[Y|X],
        r(X) = E[D|X],
        m(X) = E[Z|X].

    Returns:
        theta, se, ci_low, ci_high, plus orthogonalized variables.
    """
    Y = np.asarray(Y, dtype=float)
    D = np.asarray(D, dtype=int)
    Z = np.asarray(Z, dtype=int)
    X = np.asarray(X, dtype=float)

    n = len(Y)

    g_hat = np.zeros(n)   # E[Y|X]
    r_hat = np.zeros(n)   # E[D|X]
    m_hat = np.zeros(n)   # E[Z|X]

    reg_g, _ = make_learners(learner_g, random_state=random_state)
    _, clf_r = make_learners(learner_r, random_state=random_state)
    _, clf_m = make_learners(learner_m, random_state=random_state)

    gkf = GroupKFold(n_splits=K, shuffle=True, random_state=random_state)

    for train_idx, test_idx in gkf.split(X, groups=groups): # Pass groups here
        X_train, X_test = X[train_idx], X[test_idx]

        # g(X) = E[Y|X]
        g_model = clone(reg_g).fit(X_train, Y[train_idx])
        g_hat[test_idx] = g_model.predict(X_test)

        # r(X) = E[D|X]
        r_model = clone(clf_r).fit(X_train, D[train_idx])
        r_hat[test_idx] = r_model.predict_proba(X_test)[:, 1]

        # m(X) = E[Z|X]
        m_model = clone(clf_m).fit(X_train, Z[train_idx])
        m_hat[test_idx] = m_model.predict_proba(X_test)[:, 1]

    # Orthogonalized variables
    Y_tilde = Y - g_hat
    D_tilde = D - r_hat
    Z_tilde = Z - m_hat

    # PLIV ratio
    numer = np.mean(Z_tilde * Y_tilde)
    denom = np.mean(Z_tilde * D_tilde)

    theta_hat = numer / denom

    # Influence score
    psi_hat = (Y_tilde - theta_hat * D_tilde) * Z_tilde

    var_hat = (np.mean(psi_hat ** 2) / (denom ** 2)) / n
    se_hat = np.sqrt(var_hat)

    ci_low = theta_hat - 1.96 * se_hat
    ci_high = theta_hat + 1.96 * se_hat

    return {
        "theta_hat": theta_hat,
        "se_hat": se_hat,
        "ci_low": ci_low,
        "ci_high": ci_high,
        "numerator": numer,
        "denominator": denom,
        "Y_tilde": Y_tilde,
        "D_tilde": D_tilde,
        "Z_tilde": Z_tilde
    }


# ---------------------------------------------------
# 4. Run the estimator
# ---------------------------------------------------

# ---------------------------------------------------
# 4.1 PLIV-DML LATE LASSO
# ---------------------------------------------------

res_lasso = pliv_dml_late_conf(
    Y=Y,
    D=D,
    Z=Z,
    X=X,
    groups=groups, # Pass groups here
    K=5,
    learner_g="lasso",
    learner_r="lasso",
    learner_m="lasso",
    random_state=123
)

#print("DML-PLIV estimate of LATE")
#print(f"Theta (LATE): {res['theta_hat']:.3f}")
#print(f"SE:           {res['se_hat']:.3f}")
#print(f"95% CI:       [{res['ci_low']:.3f}, {res['ci_high']:.3f}]")
#print(f"Numerator:    {res['numerator']:.6f}")
#print(f"Denominator:  {res['denominator']:.6f}")


# ---------------------------------------------------
# 4.2 PLIV-DML LATE RANDOM FOREST
# ---------------------------------------------------
res_rf = pliv_dml_late_conf(
    Y=Y,
    D=D,
    Z=Z,
    X=X,
    groups=groups, # Pass groups here
    K=5,
    learner_g="rf",
    learner_r="rf",
    learner_m="rf",
    random_state=123
)

#print("DML-PLIV estimate of LATE")
#print(f"Theta (LATE): {res['theta_hat']:.3f}")
#print(f"SE:           {res['se_hat']:.3f}")
#print(f"95% CI:       [{res['ci_low']:.3f}, {res['ci_high']:.3f}]")
#print(f"Numerator:    {res['numerator']:.6f}")
#print(f"Denominator:  {res['denominator']:.6f}")

# ---------------------------------------------------
# 4.3 PLIV-DML LATE NEURAL NETWORK
# ---------------------------------------------------
res_mlp = pliv_dml_late_conf(
    Y=Y,
    D=D,
    Z=Z,
    X=X,
    groups=groups, # Pass groups here
    K=5,
    learner_g="mlp",
    learner_r="mlp",
    learner_m="mlp",
    random_state=123
)

#print("DML-PLIV estimate of LATE")
#print(f"Theta (LATE): {res['theta_hat']:.3f}")
#print(f"SE:           {res['se_hat']:.3f}")
#print(f"95% CI:       [{res['ci_low']:.3f}, {res['ci_high']:.3f}]")
#print(f"Numerator:    {res['numerator']:.6f}")
#print(f"Denominator:  {res['denominator']:.6f}")


results = pd.DataFrame([
    {
        "method": "DML-PLIV (Lasso)",
        "LATE": res_lasso["theta_hat"],
        "SE": res_lasso["se_hat"],
        "CI Low": res_lasso["ci_low"],
        "CI High": res_lasso["ci_high"],
    },
    {
        "method": "DML-PLIV (Random Forest)",
        "LATE": res_rf["theta_hat"],
        "SE": res_rf["se_hat"],
        "CI Low": res_rf["ci_low"],
        "CI High": res_rf["ci_high"],
    },
    {
        "method": "DML-PLIV (Neural Network)",
        "LATE": res_mlp["theta_hat"],
        "SE": res_mlp["se_hat"],
        "CI Low": res_mlp["ci_low"],
        "CI High": res_mlp["ci_high"],
    }
])

print(results.round(3))
```

## Fully Non-Parametric DML



```python
  import numpy as np
  from dataclasses import dataclass
  from typing import Optional, Dict, Any

  from sklearn.model_selection import KFold
  from sklearn.pipeline import Pipeline
  from sklearn.preprocessing import StandardScaler
  from sklearn.linear_model import LassoCV, LogisticRegressionCV
  from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
  from sklearn.neural_network import MLPRegressor, MLPClassifier


  # ============================================================
  # Result container
  # ============================================================

  @dataclass
  class LATEDMLResult:
      late: float
      se: float
      ci_lower: float
      ci_upper: float
      mean_delta_y: float
      mean_delta_d: float
      delta_y_hat: np.ndarray
      delta_d_hat: np.ndarray
      mu1_hat: np.ndarray
      mu0_hat: np.ndarray
      p1_hat: np.ndarray
      p0_hat: np.ndarray
      pi_hat: np.ndarray
      influence_hat: np.ndarray
      settings: Dict[str, Any]


  # ============================================================
  # Utilities
  # ============================================================

  def _to_numpy_1d(x, name: str) -> np.ndarray:
      arr = np.asarray(x)
      if arr.ndim != 1:
          arr = arr.ravel()
      if arr.ndim != 1:
          raise ValueError(f"{name} must be one-dimensional after flattening.")
      return arr


  def _to_numpy_2d(x, name: str) -> np.ndarray:
      arr = np.asarray(x)
      if arr.ndim == 1:
          arr = arr.reshape(-1, 1)
      if arr.ndim != 2:
          raise ValueError(f"{name} must be a 2D array or coercible to 2D.")
      return arr


  def _validate_inputs(Y, D, Z, X):
      Y = _to_numpy_1d(Y, "Y").astype(float)
      D = _to_numpy_1d(D, "D").astype(int)
      Z = _to_numpy_1d(Z, "Z").astype(int)
      X = _to_numpy_2d(X, "X").astype(float)

      n = len(Y)
      if len(D) != n or len(Z) != n or X.shape[0] != n:
          raise ValueError("Y, D, Z, and X must have the same number of observations.")

      if not np.all(np.isin(np.unique(D), [0, 1])):
          raise ValueError("D must be binary and take values in {0,1}.")
      if not np.all(np.isin(np.unique(Z), [0, 1])):
          raise ValueError("Z must be binary and take values in {0,1}.")

      return Y, D, Z, X


  # ============================================================
  # Learner factories
  # ============================================================

  def make_regressor(
      learner: str,
      random_state: Optional[int] = None,
      params: Optional[Dict[str, Any]] = None
  ):
      """
      Regression learner for continuous nuisances, such as:
      - mu_z(X) = E[Y | Z=z, X]
      """
      learner = learner.lower()
      params = {} if params is None else dict(params)

      if learner == "lasso":
          cv = params.pop("cv", 5)
          max_iter = params.pop("max_iter", 50000)
          return Pipeline([
              ("scaler", StandardScaler()),
              ("model", LassoCV(cv=cv, max_iter=max_iter,
                                random_state=random_state, **params))
          ])

      elif learner in ["rf", "random_forest", "randomforest"]:
          default_params = {
              "n_estimators": 30000,
              "min_samples_leaf": 5,
              "random_state": random_state,
              "n_jobs": -1,
          }
          default_params.update(params)
          return RandomForestRegressor(**default_params)

      elif learner in ["nn", "mlp", "neural_network", "neuralnet"]:
          default_params = {
              "hidden_layer_sizes": (16, 8),
              "activation": "relu",
              "solver": "adam",
              "alpha": 1e-4,
              "learning_rate": "adaptive",
              "max_iter": 20000,
              "early_stopping": True,
              "random_state": random_state,
          }
          default_params.update(params)
          return Pipeline([
              ("scaler", StandardScaler()),
              ("model", MLPRegressor(**default_params))
          ])

      else:
          raise ValueError("Unknown regression learner. Use 'lasso', 'rf', or 'nn'.")


  def make_classifier(
      learner: str,
      random_state: Optional[int] = None,
      params: Optional[Dict[str, Any]] = None
  ):
      """
      Classification learner for binary nuisances, such as:
      - p_z(X) = E[D | Z=z, X] = P(D=1 | Z=z, X)
      - pi(X) = P(Z=1 | X)
      """
      learner = learner.lower()
      params = {} if params is None else dict(params)

      if learner == "lasso":
          cv = params.pop("cv", 5)
          solver = params.pop("solver", "saga")
          max_iter = params.pop("max_iter", 500000)
          Cs = params.pop("Cs", 10)

          return Pipeline([
              ("scaler", StandardScaler()),
              ("model", LogisticRegressionCV(
                  penalty="l1",
                  solver=solver,
                  cv=cv,
                  Cs=Cs,
                  scoring="neg_log_loss",
                  max_iter=max_iter,
                  random_state=random_state,
                  **params
              ))
          ])

      elif learner in ["rf", "random_forest", "randomforest"]:
          default_params = {
              "n_estimators": 30000,
              "min_samples_leaf": 5,
              "random_state": random_state,
              "n_jobs": -1,
          }
          default_params.update(params)
          return RandomForestClassifier(**default_params)

      elif learner in ["nn", "mlp", "neural_network", "neuralnet"]:
          default_params = {
              "hidden_layer_sizes": (16, 8),
              "activation": "relu",
              "solver": "adam",
              "alpha": 1e-2,
              "learning_rate": "adaptive",
              "max_iter": 20000,
              "early_stopping": True,
              "random_state": random_state,
          }
          default_params.update(params)
          return Pipeline([
              ("scaler", StandardScaler()),
              ("model", MLPClassifier(**default_params))
          ])

      else:
          raise ValueError("Unknown classifier learner. Use 'lasso', 'rf', or 'nn'.")


  def _predict_proba_one(model, X):
      if hasattr(model, "predict_proba"):
          return model.predict_proba(X)[:, 1]
      raise ValueError("Binary learner must support predict_proba().")


  # ============================================================
  # Main estimator
  # ============================================================

  def late_dml_conf(
      Y,
      D,
      Z,
      X,
      K: int = 5,
      learner_mu: str = "lasso",
      learner_p: str = "lasso",
      learner_pi: str = "lasso",
      mu_params: Optional[Dict[str, Any]] = None,
      p_params: Optional[Dict[str, Any]] = None,
      pi_params: Optional[Dict[str, Any]] = None,
      clip_prob: float = 1e-3,
      random_state: Optional[int] = 123,
  ):
      """
      Fully nonparametric DML estimator of the LATE using a cross-fitted
      conditional Wald estimand.

      Target:
          tau_LATE = E[Delta_Y(X)] / E[Delta_D(X)]
      where
          Delta_Y(X) = E[Y | Z=1, X] - E[Y | Z=0, X]
          Delta_D(X) = E[D | Z=1, X] - E[D | Z=0, X]

      Key distinction from PLIV-DML:
      --------------------------------
      This estimator does NOT impose a partially linear structural model like

          Y = theta D + g(X) + U
          D = m(Z,X) + V

      Instead, it relies directly on IV potential-outcomes assumptions:
      relevance, conditional independence, exclusion, and monotonicity.
      The target is explicitly the LATE for compliers, identified through a
      nonparametric conditional Wald formula.
      """
      Y, D, Z, X = _validate_inputs(Y, D, Z, X)
      n = len(Y)

      if K < 2:
          raise ValueError("K must be at least 2.")
      if K > n:
          raise ValueError("K cannot exceed sample size.")
      if not (0 < clip_prob < 0.5):
          raise ValueError("clip_prob must lie in (0, 0.5).")

      mu1_hat = np.empty(n, dtype=float)
      mu0_hat = np.empty(n, dtype=float)
      p1_hat = np.empty(n, dtype=float)
      p0_hat = np.empty(n, dtype=float)
      pi_hat = np.empty(n, dtype=float)

      kf = KFold(n_splits=K, shuffle=True, random_state=random_state)

      for fold_id, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
          X_train, X_test = X[train_idx], X[test_idx]
          Y_train = Y[train_idx]
          D_train = D[train_idx]
          Z_train = Z[train_idx]

          z1_mask = (Z_train == 1)
          z0_mask = (Z_train == 0)

          if z1_mask.sum() == 0 or z0_mask.sum() == 0:
              raise ValueError(
                  f"Fold {fold_id}: training data contain only one instrument state. "
                  f"Try reducing K or changing the random seed."
              )

          # ----------------------------------------------------
          # Outcome regressions by instrument state
          # mu1(X) = E[Y | Z=1, X]
          # mu0(X) = E[Y | Z=0, X]
          # ----------------------------------------------------
          mu1_model = make_regressor(
              learner=learner_mu,
              random_state=None if random_state is None else random_state + fold_id,
              params=mu_params
          )
          mu0_model = make_regressor(
              learner=learner_mu,
              random_state=None if random_state is None else random_state + 1000 + fold_id,
              params=mu_params
          )

          mu1_model.fit(X_train[z1_mask], Y_train[z1_mask])
          mu0_model.fit(X_train[z0_mask], Y_train[z0_mask])

          mu1_hat[test_idx] = mu1_model.predict(X_test)
          mu0_hat[test_idx] = mu0_model.predict(X_test)

          # ----------------------------------------------------
          # Treatment regressions by instrument state
          # p1(X) = E[D | Z=1, X]
          # p0(X) = E[D | Z=0, X]
          # Since D is binary, estimate via classification probabilities
          # ----------------------------------------------------
          p1_model = make_classifier(
              learner=learner_p,
              random_state=None if random_state is None else random_state + 2000 + fold_id,
              params=p_params
          )
          p0_model = make_classifier(
              learner=learner_p,
              random_state=None if random_state is None else random_state + 3000 + fold_id,
              params=p_params
          )

          # p1(X) = E[D | Z=1, X]
          d_z1 = D_train[z1_mask]
          if np.unique(d_z1).size < 2:
              # Degenerate subgroup: all treated or all untreated
              p1_hat[test_idx] = float(d_z1[0])
          else:
              p1_model.fit(X_train[z1_mask], d_z1)
              p1_hat[test_idx] = _predict_proba_one(p1_model, X_test)

          # p0(X) = E[D | Z=0, X]
          d_z0 = D_train[z0_mask]
          if np.unique(d_z0).size < 2:
              p0_hat[test_idx] = float(d_z0[0])
          else:
              p0_model.fit(X_train[z0_mask], d_z0)
              p0_hat[test_idx] = _predict_proba_one(p0_model, X_test)


          # ----------------------------------------------------
          # Instrument propensity
          # pi(X) = P(Z=1 | X)
          # Included for documentation and optional diagnostics
          # ----------------------------------------------------
          pi_model = make_classifier(
              learner=learner_pi,
              random_state=None if random_state is None else random_state + 4000 + fold_id,
              params=pi_params
          )
          pi_model.fit(X_train, Z_train)
          pi_hat[test_idx] = _predict_proba_one(pi_model, X_test)

      # Probability clipping for stability
      p1_hat = np.clip(p1_hat, clip_prob, 1.0 - clip_prob)
      p0_hat = np.clip(p0_hat, clip_prob, 1.0 - clip_prob)
      pi_hat = np.clip(pi_hat, clip_prob, 1.0 - clip_prob)

      # Conditional Wald components
      delta_y_hat = mu1_hat - mu0_hat
      delta_d_hat = p1_hat - p0_hat

      mean_delta_y = np.mean(delta_y_hat)
      mean_delta_d = np.mean(delta_d_hat)

      if np.isclose(mean_delta_d, 0.0):
          raise ValueError(
              "The estimated first-stage Wald denominator is numerically close to zero. "
              "This suggests a weak instrument or poor first-stage fit."
          )

      late_hat = mean_delta_y / mean_delta_d

      # --------------------------------------------------------
      # Delta-method style influence approximation
      #
      # tau = EY / ED
      # with EY = E[Delta_Y(X)], ED = E[Delta_D(X)]
      #
      # Approximate influence:
      # IF_i = [ (Delta_Y_i - EY) - tau * (Delta_D_i - ED) ] / ED
      # --------------------------------------------------------
      influence_hat = (delta_y_hat - mean_delta_y - late_hat * (delta_d_hat - mean_delta_d)) / mean_delta_d

      var_hat = np.mean(influence_hat ** 2) / n
      se_hat = np.sqrt(var_hat)

      ci_lower = late_hat - 1.96 * se_hat
      ci_upper = late_hat + 1.96 * se_hat

      settings = {
          "K": K,
          "learner_mu": learner_mu,
          "learner_p": learner_p,
          "learner_pi": learner_pi,
          "mu_params": mu_params,
          "p_params": p_params,
          "pi_params": pi_params,
          "clip_prob": clip_prob,
          "random_state": random_state,
          "estimand": "Fully nonparametric conditional Wald LATE",
      }

      return LATEDMLResult(
          late=float(late_hat),
          se=float(se_hat),
          ci_lower=float(ci_lower),
          ci_upper=float(ci_upper),
          mean_delta_y=float(mean_delta_y),
          mean_delta_d=float(mean_delta_d),
          delta_y_hat=delta_y_hat,
          delta_d_hat=delta_d_hat,
          mu1_hat=mu1_hat,
          mu0_hat=mu0_hat,
          p1_hat=p1_hat,
          p0_hat=p0_hat,
          pi_hat=pi_hat,
          influence_hat=influence_hat,
          settings=settings,
      )


  # ============================================================
  # Summary printer
  # ============================================================

  def summarize_late_dml_result(res: LATEDMLResult):
      print("Fully Nonparametric DML LATE Estimate")
      print("-------------------------------------")
      print(f"LATE estimate : {res.late:.6f}")
      print(f"Std. Error    : {res.se:.6f}")
      print(f"95% CI        : [{res.ci_lower:.6f}, {res.ci_upper:.6f}]")
      print(f"E[Delta_Y(X)] : {res.mean_delta_y:.6f}")
      print(f"E[Delta_D(X)] : {res.mean_delta_d:.6f}")
      print()
      print("Settings")
      for k, v in res.settings.items():
          print(f"{k}: {v}")


  # ============================================================
  # Implemented with Sample data
  # ============================================================

  if __name__ == "__main__":

      # Select variables needed for Part 1
      features_base = ['field_size', 'total_labor_day', 'ha_harvest', 't2m',
                 'rh2m', 'prectotcorr', 'inorg_fert_kg', 'org_fert_kg',
                 'pest_kg', 'herb_kg']

      cols_needed = ["quant_harv_kg", "treatment_fatalities", "military_presence", 'conflict_event', 'lga', 'year'] + features_base

      # Keep only complete cases
      df = sorghum_data[cols_needed].dropna().copy()

      cols_to_demean = ["quant_harv_kg","treatment_fatalities", 'conflict_event', "military_presence"] + features_base

      df = two_way_demean (
        df,
        cols=cols_to_demean,
        unit_col="lga",
        time_col="year"
    )

      # Outcome
      Y = df["quant_harv_kg"].to_numpy(dtype=float)

      # Treatment
      D = df["conflict_event"].to_numpy(dtype=int)

      # Elligibility
      Z = df["military_presence"].to_numpy(dtype=int)

      X_cols_twfe = [f"{x}_twfe" for x in features_base]
      X = df[X_cols_twfe].to_numpy(dtype=float)

      groups = (df['lga'].astype(str) + '_' + df['year'].astype(str)).to_numpy() # Re-defining groups as 1D array of strings


      print("Shapes:")
      print("Y:", Y.shape)
      print("D:", D.shape)
      print("Z:", Z.shape)
      print("X:", X.shape)
      print("Participation share:", D.mean())
      print("Elligibility share:", Z.mean())


      # Lasso version
      res_lasso = late_dml_conf(
          Y=Y,
          D=D,
          Z=Z,
          X=X,
          K=5,
          learner_mu="lasso",
          learner_p="lasso",
          learner_pi="lasso",
          clip_prob=1e-2,
          random_state=123
      )
      #summarize_late_dml_result(res_lasso)
      #print()

      # Random Forest version
      res_rf = late_dml_conf(
          Y=Y,
          D=D,
          Z=Z,
          X=X,
          K=5,
          learner_mu="rf",
          learner_p="rf",
          learner_pi="rf",
          mu_params={"n_estimators": 5000, "min_samples_leaf": 5},
          p_params={"n_estimators": 5000, "min_samples_leaf": 5},
          pi_params={"n_estimators": 5000, "min_samples_leaf": 5},
          clip_prob=1e-2,
          random_state=123
      )
      #summarize_late_dml_result(res_rf)
      #print()

      # Neural network version
      res_nn = late_dml_conf(
          Y=Y,
          D=D,
          Z=Z,
          X=X,
          K=5,
          learner_mu="nn",
          learner_p="nn",
          learner_pi="nn",
          mu_params={"hidden_layer_sizes": (16, 8), "max_iter": 20000},
          p_params={"hidden_layer_sizes": (16, 8), "max_iter": 20000},
          pi_params={"hidden_layer_sizes": (16, 8), "max_iter": 20000},
          clip_prob=1e-2,
          random_state=123
      )
      #summarize_late_dml_result(res_nn)


      results = pd.DataFrame([
      {
          "method": "Non-Par. DML (Lasso)",
          "LATE": res_lasso.late,
          "SE": res_lasso.se,
          "CI Low": res_lasso.ci_lower,
          "CI High": res_lasso.ci_upper,
      },
      {
          "method": "Non-Par. DML (Random Forest)",
          "LATE": res_rf.late,
          "SE": res_rf.se,
          "CI Low": res_rf.ci_lower,
          "CI High": res_rf.ci_upper,
      },
      {
          "method": "Non-Par. DML (Neural Network)",
          "LATE": res_nn.late,
          "SE": res_nn.se,
          "CI Low": res_nn.ci_lower,
          "CI High": res_nn.ci_upper,
      }
  ])

  print(results.round(3))
```
